```python
import pandas as pd
from pathlib import Path

data_path = Path("../data")
csv_files = list(data_path.glob("*.csv"))

df = pd.read_csv(csv_files[0])

punjab_df = df[df['SUBDIVISION'] == 'Punjab'].copy()

punjab_df = punjab_df.sort_values('YEAR').reset_index(drop=True)

print("Punjab records:", len(punjab_df))
print("Years:", punjab_df['YEAR'].min(), "-", punjab_df['YEAR'].max())
```

```python
forecast_df = punjab_df.copy()

forecast_df['TARGET_YEAR'] = forecast_df['YEAR'].shift(-1)
forecast_df['TARGET_ANNUAL'] = forecast_df['ANNUAL'].shift(-1)

forecast_df = forecast_df.dropna().reset_index(drop=True)

forecast_df[['YEAR', 'ANNUAL', 'TARGET_YEAR', 'TARGET_ANNUAL']].head()
```

```python
monthly_features = [
    'JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN',
    'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC'
]

X_forecast = forecast_df[monthly_features]
y_forecast = forecast_df['TARGET_ANNUAL']

split_index = int(len(X_forecast) * 0.8)

X_train_forecast = X_forecast.iloc[:split_index]
X_test_forecast = X_forecast.iloc[split_index:]

y_train_forecast = y_forecast.iloc[:split_index]
y_test_forecast = y_forecast.iloc[split_index:]

print("Training data:", X_train_forecast.shape)
print("Testing data:", X_test_forecast.shape)
print("Training target years:", forecast_df['TARGET_YEAR'].iloc[:split_index].min(), 
      "-", forecast_df['TARGET_YEAR'].iloc[:split_index].max())
print("Testing target years:", forecast_df['TARGET_YEAR'].iloc[split_index:].min(), 
      "-", forecast_df['TARGET_YEAR'].iloc[split_index:].max())
```

```python
model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

model.fit(X_train_forecast, y_train_forecast)

print("Model training completed!")
```

```python
y_pred_forecast = model.predict(X_test_forecast)

print("Predictions completed!")
print(y_pred_forecast[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test_forecast, y_pred_forecast)
mse = mean_squared_error(y_test_forecast, y_pred_forecast)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_forecast, y_pred_forecast)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R² Score:", round(r2, 3))
```

```python
baseline_pred = np.full(
    len(y_test_forecast),
    y_train_forecast.mean()
)

baseline_mae = mean_absolute_error(
    y_test_forecast,
    baseline_pred
)

baseline_r2 = r2_score(
    y_test_forecast,
    baseline_pred
)

print("Baseline MAE:", round(baseline_mae, 2))
print("Baseline R2:", round(baseline_r2, 3))
print("Random Forest MAE:", round(mae, 2))
print("Random Forest R2:", round(r2, 3))
```

```python
monthly_df = punjab_df[['YEAR'] + monthly_features].melt(
    id_vars='YEAR',
    var_name='MONTH',
    value_name='RAINFALL'
)

monthly_df.head(15)
```

```python
month_mapping = {
    'JAN': 1, 'FEB': 2, 'MAR': 3,
    'APR': 4, 'MAY': 5, 'JUN': 6,
    'JUL': 7, 'AUG': 8, 'SEP': 9,
    'OCT': 10, 'NOV': 11, 'DEC': 12
}

monthly_df['MONTH_NUM'] = monthly_df['MONTH'].map(month_mapping)

monthly_df.head(15)
```

```python
monthly_df = monthly_df.sort_values(
    ['YEAR', 'MONTH_NUM']
).reset_index(drop=True)

monthly_df.head(15)
```

```python
monthly_df['DATE'] = pd.to_datetime({
    'year': monthly_df['YEAR'],
    'month': monthly_df['MONTH_NUM'],
    'day': 1
})

monthly_df.head(15)
```

```python
monthly_df['PREVIOUS_MONTH_RAINFALL'] = monthly_df['RAINFALL'].shift(1)
monthly_df['PREVIOUS_YEAR_RAINFALL'] = monthly_df['RAINFALL'].shift(12)

monthly_df = monthly_df.dropna().reset_index(drop=True)

monthly_df.head()
```

```python
monthly_features_model = [
    'MONTH_NUM',
    'PREVIOUS_MONTH_RAINFALL',
    'PREVIOUS_YEAR_RAINFALL'
]

X_monthly = monthly_df[monthly_features_model]
y_monthly = monthly_df['RAINFALL']

split_index_monthly = int(len(X_monthly) * 0.8)

X_train_monthly = X_monthly.iloc[:split_index_monthly]
X_test_monthly = X_monthly.iloc[split_index_monthly:]

y_train_monthly = y_monthly.iloc[:split_index_monthly]
y_test_monthly = y_monthly.iloc[split_index_monthly:]

print("Training records:", len(X_train_monthly))
print("Testing records:", len(X_test_monthly))
```

```python
from sklearn.ensemble import RandomForestRegressor

monthly_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

monthly_model.fit(X_train_monthly, y_train_monthly)

monthly_predictions = monthly_model.predict(X_test_monthly)

print("Monthly model trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

monthly_mae = mean_absolute_error(y_test_monthly, monthly_predictions)
monthly_mse = mean_squared_error(y_test_monthly, monthly_predictions)
monthly_rmse = np.sqrt(monthly_mse)
monthly_r2 = r2_score(y_test_monthly, monthly_predictions)

print("Monthly MAE:", round(monthly_mae, 2))
print("Monthly MSE:", round(monthly_mse, 2))
print("Monthly RMSE:", round(monthly_rmse, 2))
print("Monthly R2:", round(monthly_r2, 3))
```

```python
import joblib
from pathlib import Path

model_path = Path("../models/monthly_rainfall_model.pkl")
joblib.dump(monthly_model, model_path)

print("Monthly model saved successfully!")
```
